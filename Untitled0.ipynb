```python
import sqlite3
import pandas as pd
import plotly.express as px
import math
from IPython.display import display, HTML

DB_PATH = '/content/usda_food.db'
conn = sqlite3.connect(DB_PATH)

# ============================================================
# 1. СПРАВОЧНИК ПЕРЕВОДОВ И ДОБАВОК
# ============================================================
RU_NAMES = {
    'Cereals, oats, regular and quick, not fortified, dry': 'Овсяные хлопья',
    'Egg, whole, cooked, hard-boiled': 'Яйцо куриное',
    'Butter, without salt': 'Масло сливочное',
    'Cheese, cottage, lowfat, 1% milkfat': 'Творог 1%',
    'Apricots, dried, sulfured, uncooked': 'Курага',
    'Chicken, broilers or fryers, breast, meat only, cooked, roasted': 'Куриная грудка',
    'Buckwheat groats, roasted, cooked': 'Гречка (готовая)',
    'Carrots, cooked, boiled, drained, without salt': 'Морковь отварная',
    'Oil, olive, salad or cooking': 'Масло оливковое',
    'Kefir, lowfat, plain, LIFEWAY': 'Кефир 1%',
    'Bananas, raw': 'Банан',
    'Potatoes, boiled, cooked without skin, flesh, without salt': 'Картофель отварной',
    'Broccoli, cooked, boiled, drained, without salt': 'Брокколи отварная',
    'Cheese, cheddar': 'Сыр Чеддер',
}

# Стоимость БАДов в день (Компливит + D3 + Омега-3 + Магний)
SUPPLEMENTS_COST_PER_DAY = 35.0

RDA = {
    'Vitamin A, RAE': 900,            # мкг
    'Vitamin C': 100,                 # мг (было 90)
    'Vitamin D (D2 + D3)': 15,        # мкг
    'Vitamin E (alpha-tocopherol)': 15, # мг
    'Vitamin K (phylloquinone)': 120, # мкг
    'Thiamin': 1.5,                   # мг (было 1.2)
    'Riboflavin': 1.8,                # мг (было 1.3)
    'Niacin': 20,                     # мг (было 16)
    'Vitamin B-6': 2.0,               # мг (было 1.3)
    'Folate, DFE': 400,               # мкг
    'Vitamin B-12': 3.0,              # мкг (было 2.4)
    'Calcium, Ca': 1000,              # мг
    'Magnesium, Mg': 420,             # мг
    'Potassium, K': 3500,             # мг
    'Iron, Fe': 10,                   # мг (было 8)
    'Zinc, Zn': 12,                   # мг (было 11)
}
# ============================================================
# ТВОИ РЕАЛЬНЫЕ БАДЫ (Суммарно за день)
# ============================================================
SUPPLEMENTS = {
    'Vitamin A, RAE': 800,           # из Компливита (0.8 мг)
    'Vitamin C': 25,                 # из Компливита
    'Vitamin D (D2 + D3)': 50,       # из капсул D3 (2000 МЕ)
    'Vitamin E (alpha-tocopherol)': 3.0, # из Компливита
    'Thiamin': 1.0,                  # B1 из Компливита
    'Riboflavin': 2.62,              # B2 из Компливита (1.27) + из Магния (1.35)
    'Niacin': 7.5,                   # B3 из Компливита (Никотинамид)
    'Vitamin B-6': 3.35,             # B6 из Компливита (2.0) + из Магния (1.35)
    'Folate, DFE': 100,              # B9 из Компливита (0.1 мг = 100 мкг)
    'Vitamin B-12': 1.5,             # из Компливита
    'Calcium, Ca': 0,                # нет в твоих БАДах
    'Magnesium, Mg': 130,            # 60 из Компливита + 70 из Хелата
    'Potassium, K': 0,               # нет в БАДах (только из еды)
    'Iron, Fe': 5.0,                 # из Компливита
    'Zinc, Zn': 7.5,                 # 2.5 из Компливита + 5.0 из Магния
}

# ============================================================
# 2. ВВОД ЦЕН (Только тут!)
# ============================================================
# Указывай ТОЛЬКО вес пачки (в граммах) и её реальную цену в магазине.
# Названия продуктов должны точно совпадать с теми, что в MENUS.
PACKAGES = {
    'Cereals, oats, regular and quick, not fortified, dry': {'вес': 1000, 'цена': 100},
    'Egg, whole, cooked, hard-boiled': {'вес': 600, 'цена': 120},         # Лоток 10 шт
    'Butter, without salt': {'вес': 180, 'цена': 200},                    # Пачка 180г
    'Cheese, cottage, lowfat, 1% milkfat': {'вес': 300, 'цена': 90},      # Пачка 300г
    'Apricots, dried, sulfured, uncooked': {'вес': 200, 'цена': 80},      # Пачка 200г
    'Chicken, broilers or fryers, breast, meat only, cooked, roasted': {'вес': 900, 'цена': 320}, # Подложка 0.9кг
    'Buckwheat groats, roasted, cooked': {'вес': 800, 'цена': 75},        # Пачка 800г
    'Carrots, cooked, boiled, drained, without salt': {'вес': 1000, 'цена': 30}, # 1 кг
    'Oil, olive, salad or cooking': {'вес': 500, 'цена': 500},            # Бутылка 0.5л
    'Kefir, lowfat, plain, LIFEWAY': {'вес': 900, 'цена': 80},            # Пакет 0.9л
    'Bananas, raw': {'вес': 1000, 'цена': 90},                            # 1 кг
    'Potatoes, boiled, cooked without skin, flesh, without salt': {'вес': 2000, 'цена': 60}, # Сетка 2 кг
    'Broccoli, cooked, boiled, drained, without salt': {'вес': 400, 'цена': 100}, # Заморозка
    'Cheese, cheddar': {'вес': 200, 'цена': 120},                         # Кусок 200г
}

# ============================================================
# 3. АВТОМАТИЧЕСКИЙ РАСЧЕТ ЦЕНЫ ЗА 100 ГРАММ
# ============================================================
PRICES = {}
for name, pack in PACKAGES.items():
    if pack['вес'] > 0:
        PRICES[name] = (pack['цена'] / pack['вес']) * 100

# ============================================================
# 4. ТВОИ ВАРИАНТЫ МЕНЮ (Добавляй сюда новые)
# ============================================================
MENUS = {
    "Базовое меню (текущее)": [
        ('Cereals, oats, regular and quick, not fortified, dry', 60),
        ('Egg, whole, cooked, hard-boiled', 100),
        ('Butter, without salt', 10),
        ('Cheese, cottage, lowfat, 1% milkfat', 100),
        ('Apricots, dried, sulfured, uncooked', 50),
        ('Chicken, broilers or fryers, breast, meat only, cooked, roasted', 120),
        ('Buckwheat groats, roasted, cooked', 150),
        ('Carrots, cooked, boiled, drained, without salt', 100),
        ('Oil, olive, salad or cooking', 10),
        ('Kefir, lowfat, plain, LIFEWAY', 200),
        ('Bananas, raw', 100),
        ('Chicken, broilers or fryers, breast, meat only, cooked, roasted', 100),
        ('Potatoes, boiled, cooked without skin, flesh, without salt', 300),
        ('Broccoli, cooked, boiled, drained, without salt', 100),
        ('Cheese, cheddar', 30),
        ('Butter, without salt', 10),
        ('Cheese, cottage, lowfat, 1% milkfat', 100),
    ],

    "Эконом-вариант (тест)": [
        ('Cereals, oats, regular and quick, not fortified, dry', 60),
        ('Egg, whole, cooked, hard-boiled', 150),
        ('Butter, without salt', 10),
        ('Cheese, cottage, lowfat, 1% milkfat', 200),
        ('Chicken, broilers or fryers, breast, meat only, cooked, roasted', 200),
        ('Buckwheat groats, roasted, cooked', 200),
        ('Carrots, cooked, boiled, drained, without salt', 100),
        ('Oil, olive, salad or cooking', 10),
        ('Potatoes, boiled, cooked without skin, flesh, without salt', 400),
    ]
}

# ============================================================
# 5. ФУНКЦИИ РАСЧЕТА
# ============================================================
def get_nutrients(conn, name):
    q = """SELECT n.name AS nutrient_name, fn.amount FROM food f
           JOIN food_nutrient fn ON f.fdc_id = fn.fdc_id
           JOIN nutrient n ON fn.nutrient_id = n.id
           WHERE f.description = ?"""
    df = pd.read_sql_query(q, conn, params=(name,))
    return dict(zip(df['nutrient_name'], df['amount']))

def process_menu(menu_name, products):
    # --- А) Экономика ---
    monthly_needs = {}
    for name, grams in products:
        monthly_needs[name] = monthly_needs.get(name, 0) + (grams * 30)

    shopping_list = []
    actual_cost = 0
    theo_cost = 0

    for name, total_grams in monthly_needs.items():
        price_100g = PRICES.get(name, 0)
        t_cost = (total_grams / 100) * price_100g
        theo_cost += t_cost

        pack = PACKAGES.get(name)
        if pack:
            packs_needed = math.ceil(total_grams / pack['вес'])
            r_cost = packs_needed * pack['цена']
            actual_cost += r_cost

            shopping_list.append({
                'Продукт': RU_NAMES.get(name, name),
                'Нужно (г)': round(total_grams),
                'Пачек (шт)': packs_needed,
                'Чек (руб)': r_cost
            })

    actual_cost += (SUPPLEMENTS_COST_PER_DAY * 30)
    shopping_df = pd.DataFrame(shopping_list).sort_values('Чек (руб)', ascending=False)

    # --- Б) Нутриенты ---
    total_nutr = {k: 0 for k in RDA}
    for name, grams in products:
        nutr = get_nutrients(conn, name)
        if not nutr: continue
        factor = grams / 100.0
        for k in RDA:
            total_nutr[k] += nutr.get(k, 0) * factor

    for k in RDA:
        total_nutr[k] += SUPPLEMENTS.get(k, 0)

    results_rows = []
    for nutrient, rda_value in RDA.items():
        value = total_nutr.get(nutrient, 0)
        percent = (value / rda_value) * 100 if rda_value else 0
        status = '✅ Норма' if percent >= 100 else '⚠️ Дефицит' if percent >= 50 else '🔴 Сильный дефицит'
        results_rows.append({'Нутриент': nutrient, 'Процент': round(percent, 1), 'Статус': status})
    results_df = pd.DataFrame(results_rows).sort_values('Процент')

    # --- В) Графики ---
    fig_pie = px.pie(shopping_df, values='Чек (руб)', names='Продукт',
                     title=f'Расходы: {menu_name}', hole=0.4)
    pie_html = fig_pie.to_html(full_html=False, include_plotlyjs=False)

    fig_bar = px.bar(
        results_df, x='Процент', y='Нутриент', orientation='h', color='Статус',
        color_discrete_map={'✅ Норма': '#2ecc71', '⚠️ Дефицит': '#f1c40f', '🔴 Сильный дефицит': '#e74c3c'},
        title='Микронутриенты'
    )
    fig_bar.update_traces(texttemplate='%{x:.1f}%', textposition='outside')
    fig_bar.update_layout(xaxis_range=[0, max(120, results_df['Процент'].max() * 1.15)], height=500)
    fig_bar.add_vline(x=100, line_dash="dash", line_color="green")
    bar_html = fig_bar.to_html(full_html=False, include_plotlyjs=False)

    # --- Г) Сборка HTML ---
    html_content = f"""
    <p><b>Теоретическая стоимость (по граммам):</b> {theo_cost:.2f} ₽</p>
    <p><b>Разница (остатки в холодильнике):</b> {actual_cost - theo_cost:.2f} ₽</p>

    <h4>🛒 Список покупок:</h4>
    {shopping_df.to_html(index=False, classes='table table-striped')}

    <h4>📊 Расходы по категориям:</h4>
    {pie_html}

    <h4>🧬 Микронутриенты:</h4>
    {bar_html}
    """

    return actual_cost, html_content

# ============================================================
# 6. ЗАПУСК ЦИКЛА ПО ВСЕМ МЕНЮ
# ============================================================
display(HTML('<script src="https://cdn.plot.ly/plotly-latest.min.js"></script>'))

print("АНАЛИЗ ВАРИАНТОВ МЕНЮ:\n")

for menu_name, products in MENUS.items():
    total_check, content_html = process_menu(menu_name, products)

    spoiler = f"""
    <details style="margin-bottom: 20px; border: 1px solid #ccc; border-radius: 5px; padding: 10px; background-color: #f9f9f9;">
        <summary style="font-size: 18px; font-weight: bold; cursor: pointer; color: #1a73e8;">
            📁 {menu_name} — <span style="color: #e67e22;">Чек в магазине: {total_check:.2f} ₽/мес</span>
        </summary>
        <div style="margin-top: 15px; padding: 10px; background: white; border-radius: 5px;">
            {content_html}
        </div>
    </details>
    """
    display(HTML(spoiler))
```
